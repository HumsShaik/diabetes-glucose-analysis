# Project Overview

Type 1 Diabetes (T1D) is a chronic medical condition in which the body is unable to properly regulate blood glucose levels due to insufficient insulin production. Effective glucose management requires continuous monitoring of multiple physiological and behavioral factors such as insulin dosage, carbohydrate intake, physical activity, heart rate, and other wearable sensor measurements.

With the advancement of Continuous Glucose Monitoring (CGM) systems and wearable health technologies, large amounts of patient-generated physiological data are now available for analysis. Understanding relationships between these variables can help identify glucose fluctuation patterns and support future predictive healthcare applications.

---

# Problem Statement

Managing blood glucose levels in Type 1 Diabetes patients is complex because glucose variability is influenced by multiple interconnected physiological and lifestyle factors. Sudden glucose spikes or drops can lead to severe health complications if not monitored effectively.

The challenge is to analyze how different variables such as insulin administration, carbohydrate consumption, physical activity, and physiological signals impact glucose behavior over time.

# 1. Glucose Distribution Visualization
Why important?

This gives:

- first visual understanding of glucose variability
- evidence of hypo/hyperglycemia
- strong exploratory analysis section

```python
# 1. Glucose Distribution

sns.histplot(df_patient['glucose'], bins=30, kde=True)
plt.title("Glucose Distribution")
plt.xlabel("Glucose Level")
plt.ylabel("Frequency")
plt.show()
```

## Observation — Glucose Distribution

The glucose distribution shows substantial variability across the monitoring period. A large proportion of glucose readings appear concentrated in elevated glucose ranges, indicating frequent hyperglycemic conditions.

The distribution is not perfectly symmetrical and appears slightly right-skewed, suggesting the presence of extreme high glucose episodes. Lower glucose values are also present, indicating occasional hypoglycemic events.

These observations suggest that glucose regulation varies significantly over time and may be influenced by insulin administration, carbohydrate intake, and physical activity patterns.

# 2. Time-Series Glucose Plot
Why important?

This is your:

- first real healthcare time-series visualization
- core proof that the dataset is longitudinal

```python
# time-series visualization
plt.figure(figsize=(15,6))

plt.plot(df_patient['time'], df_patient['glucose'])

plt.title("Glucose Levels Over Time")
plt.xlabel("Time")
plt.ylabel("Glucose")

plt.show()
```

## Observation — Glucose Levels Over Time

The time-series visualization demonstrates continuous fluctuations in glucose levels throughout the monitoring period. Several sharp increases and decreases in glucose values are visible, indicating dynamic physiological responses over time.

Periods of sustained elevated glucose levels are observed alongside more stable intervals, suggesting variations in patient behavior, insulin response, meal intake, or activity levels.

The consistent 5-minute recording frequency provides high temporal granularity, making the dataset well-suited for longitudinal glucose trend analysis and event-response studies.
